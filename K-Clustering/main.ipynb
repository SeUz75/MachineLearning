```python

import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python

# sklearn.datasets -> library that has some datas which they get downloaded from a website.
from sklearn.datasets import make_blobs


# make_blobs generating coordinates x and y
points, cluster_indexes = make_blobs(n_samples=300,centers =4, cluster_std=0.8,random_state=0)

x=points[:,0]
y=points[:,1]

# scatter to visualise these coordinates
plt.scatter(x,y,s=50,alpha=0.7)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(points)

predicted_cluster_indexes = kmeans.predict(points)

plt.scatter(x,y,c=predicted_cluster_indexes, s=50,alpha=0.7,cmap='viridis')

centers=kmeans.cluster_centers_

plt.scatter(centers[:,0],centers[:,1],c='red',s=100)

# where i left off Chetvurta Kletka !
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=0) # with 5 clusters 
kmeans.fit(points)

predicted_cluster_indexes = kmeans.predict(points)

plt.scatter(x,y,c=predicted_cluster_indexes, s=50,alpha=0.7,cmap='viridis')

centers=kmeans.cluster_centers_

plt.scatter(centers[:,0],centers[:,1],c='red',s=100)

# where i left off Chetvurta Kletka !
```

```python
# How  to determine how many clusters do you need ?
# So here is a method
inertias=[]

for i in range (1,10):
    kmeans = KMeans(n_clusters=i,random_state=0)
    kmeans.fit(points)
    inertias.append(kmeans.inertia_)
plt.plot(range(1,10),inertias)
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')

# BRUH thats so cool so what I've learned it we are using this elbow technique to determine how many clusters we need
# and if you can see in my for loop for the DATA 'points' im using this method and we can see the elbow on the graph
# if we had different set of data the graph would be different as we as our choice of CLUSTERS ! 
```

```python
kmeans = KMeans(n_clusters=4,random_state=0)
kmeans.fit(points)
predicted_cluster_indexes = kmeans.predict(points)

plt.scatter(x,y,c=predicted_cluster_indexes, s=50,alpha=0.7,cmap='viridis')

centers = kmeans.cluster_centers_

plt.scatter(centers[:,0], centers[:,1],c='red',s=100)
```

```python
# adding a new DATA point 
import numpy as np
point = np.array([[-0.5,5]])
px=point[:,0]
py=point[:,1]

plt.scatter(x,y,c=predicted_cluster_indexes,s=50,alpha=0.7,cmap='viridis')
plt.scatter(centers[:,0],centers[:,1],c='red',s=100)
plt.scatter(px,py,c='orange',s=200)
```

```python
# in which cluster does my new point of data belong to ? 
cluster =kmeans.predict(point)[0]
print(cluster)


r=points[predicted_cluster_indexes == cluster]

plt.scatter(x,y,c=predicted_cluster_indexes,s=50,alpha=0.7,cmap='viridis')
plt.scatter(r[:,0],r[:,1],c='red',s=100)
plt.scatter(px,py,c='orange',s=200)
```

```python
# Showing the distance between my new DATA point with every cluster center
d=[]

for i in range(len(centers)):
    d.append(np.sqrt((px-centers[i][0])**2+(py-centers[i][1])**2))

print(d)

```
