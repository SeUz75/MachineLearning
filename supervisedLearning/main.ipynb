```python
import pandas as pd 
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns= iris.feature_names)
df['class'] = iris.target
df['class name'] = iris.target_names[iris['target']]
df.head()
#Loaded my data
```

```python
%matplotlib inline
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set()

sns.pairplot(df)
# Visual representation of data ! 
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=0)
# spliting my data to test and train with the help of the TRAIN_TEST_SPLIT
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
model.fit(x_train,y_train)
# Creating my module ! KNeighborsClassifier() ! 
# then training my model ! 
```

```python
model.score(x_test,y_test)
# Testing my module ! See how good it works with TEST data!
```

```python
predicted_classes = model.predict([[5.6,4.4,1.2,0.4]])
# Entering new data with these param [5.6,4.4,1.2,0.4], and predicting the output and saving it to predicted_classes var
print(predicted_classes)
# Printing the prediciton
print(iris.target_names[predicted_classes[0]])
# Checking which label had 0 for index !
```
